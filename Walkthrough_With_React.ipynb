1. Explain the potential SEO challenges and solutions associated with Single Page Applications (SPAs).

->

*   Client-side rendering (CSR) → Content loads via JS, so search  bots may miss or delay indexing
*   Same meta tags → No unique titles/descriptions per page

*   Routing issues → Pages don’t have real URLs (bad for crawling)
*   Slow load time → Heavy JS hurts SEO (Core Web Vitals)


*   Limited crawlability → Bots may not discover all pages


2. Explain the key differences and use cases between React's Client-Side Rendering (CSR) and Server-Side
Rendering (SSR). Provide examples of scenarios where each approach is advantageous, and discuss the
challenges associated with using React in both contexts.

->


*   Client-Side Rendering (CSR): The browser loads a basic HTML file and uses JavaScript (via React) to render content dynamically.
*  Server-Side Rendering (SSR): The server generates full HTML for each request and sends it to the browser, commonly using Next.js.

| Feature            | CSR             | SSR             |
| ------------------ | --------------- | --------------- |
| Rendering Location | Browser         | Server          |
| Initial Load       | Slower          | Faster          |
| SEO                | Poor            | Good            |
| Interactivity      | High after load | Needs hydration |
| Complexity         | Simpler         | More complex    |

✅ When Client-Side Rendering (CSR) is Advantageous

Using React with CSR is ideal when interactivity matters more than SEO.

📌 Scenarios
Admin dashboards (e.g., analytics panels)
SaaS platforms (CRM, project management tools)
Authenticated apps (user must log in)
Real-time apps (chat apps, live data feeds)
🎯 Why CSR works well here
SEO is not required (content is private)
UI updates frequently (state-heavy apps)
Smooth navigation without page reloads

✅ When Server-Side Rendering (SSR) is Advantageous

Using Next.js enables SSR for better SEO and performance.

📌 Scenarios
E-commerce websites (product pages)
Blogs & news sites
Marketing / landing pages
Public company websites
🎯 Why SSR works well here
Content must be indexed by search engines
Faster initial load improves user experience
Better social media previews (meta tags)

4.Create a React component called "Login" without using JSX for its definition. This component should render
an HTML structure that includes a form with input fields for a username and password. It should also have a
button labeled "Submit."

```JSX
import React from "react";

function Login() {
  return React.createElement(
    "div",
    null,
    React.createElement(
      "form",
      null,
      React.createElement("h2", null, "Login"),
      
      React.createElement("input", {
        type: "text",
        placeholder: "Username",
        name: "username"
      }),

      React.createElement("br"),

      React.createElement("input", {
        type: "password",
        placeholder: "Password",
        name: "password"
      }),

      React.createElement("br"),

      React.createElement(
        "button",
        { type: "submit" },
        "Submit"
      )
    )
  );
}

export default Login;
```
